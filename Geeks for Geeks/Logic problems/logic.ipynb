```python
#Check even or odd
n=int(input())
if n%2==0:  #if isEven(n)
    print("even")
else:
    print("odd")
```

```python
#multiplication table
n=int(input())
for i in range(1,11):
    a=n*i
    print(a)
```

```python
#sum of naturals
n=int(input())
a=0
while n>0:
    a=n+a#on formula (n(n+1)/2)
    n-=1
print(a)
```

```python
#sum of squares
n=int(input())
a=0
while n>0:
    a=(n*n)+a#on formula (n(n+1)(2n+1)/6)
    n-=1
print(a)
```

```python
#sum of digits
n=int(input())

n1=n//10
n2=n-(n1*10)
print(n1)
print(n2)

```

```python
#reverse the digits
n=int(input())

n1=n//10
n2=n-(n1*10)
print(n1)
print(n2)
print(n2*10+n1)
```

```python
#prime testing
n=int(input())
a=0
for i in range(1,n):
    if n%i==0:
        a=1
    else:
        a=0

if a==0:
    print("not prime")
else:
    print("prime")
```
