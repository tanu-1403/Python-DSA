```python
n=int(input())
s=input().strip()
for i in range(n):
    for j in s:
        if j=="(":
            s=s.strip("(")
        else:
            s=s.strip(")")

if len(s)%2:
    print("NO")
else:
    print("YES")
    
```

```python
#down with brackets
n=int(input())
for j in range(n):
    s=(input().strip())
    a=0
    result=True
    for i in s:
        if i=="(":
            a+=1
        else:
            a-=1
            
        if a<0:
            result=False
            break

    if a==0 and result:
        print("YES")
    else:
        print("NO")
        
#extra for loop iteration
```
