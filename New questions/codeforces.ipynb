```python
value=int(input("Enter a value: "))
string=""

for i in range(value):
    input_value=input("Enter list value: ")
    if input_value+string<string+input_value:
        string=input_value+string
    else:
        string=string+input_value

print(string)
```

```python
n=int(input())
s=input().strip()
for i in range(n):
    for j in s:
        if j=="(":
            s=s.strip("(")
        else:
            s=s.strip(")")

if len(s)%2:
    print("NO")
else:
    print("YES")
    
```

```python
n=int(input())
s=""
for i in range(n):
    s+=input("value:")
s=s.strip()
print("s:",s)
s1=s[0:]
print("s1:",s1)
s2=s[-1:-n-1:-1]
print("s2:",s2)
if s1==s2:
    print("YES")
else:
    print("NO")
```

```python
n=int(input())
s=""
for i in range(n):
    s+=input("value:")
s=s.strip()
print("s:",s)
a=0
result=False
for i in range(len(s)-1):
    if s[i]=="(":
        a+=1
    else:
        a-=1
        
    if a==0:
        result=True
        break

if result==True:
    print("YES")
else:
    print("NO")

```
